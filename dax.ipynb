   - `Products.csv`
   - `Customers.csv`

### **Step 2: Open Power BI Desktop**

5. Click **Open**
6. Click **Load** (not Transform)

### **Step 4: Repeat for Other Files**
1. Click **Home tab** → **Get Data** → **Text/CSV**
2. Load `Locations.csv`
3. Repeat for `Products.csv` and `Customers.csv`

## 📅 **PART 3: CREATING DATE TABLE WITH DAX**

### **Step 1: Switch to Data View**
Click the **second icon** in left sidebar

### **Step 2: Create New Table**
1. Click **Home tab**
2. Click **New Table** (in Calculations group)

![New Table Button](https://docs.microsoft.com/en-us/power-bi/transform-model/media/desktop-calculated-tables/calc-tables-new-table.png)

### **Step 3: Enter DAX Formula**
1. A formula bar appears at the top
2. Type exactly:
```dax
Dates = 
CALENDARAUTO()
```
3. Press **Enter** or click the checkmark ✓

### **Step 4: Add Date Columns**
1. With Dates table selected, click **Table tools** tab
2. Click **New Column**
3. In formula bar, type:
```dax
Year = YEAR(Dates[Date])
```
4. Press **Enter**

5. Repeat for more columns:
```dax
Month = FORMAT(Dates[Date], "MMMM")
Quarter = "Q" & QUARTER(Dates[Date])
YearMonth = FORMAT(Dates[Date], "YYYY-MM")
```

### **Step 5: Create Relationship**
1. Go back to **Model view** (third icon)
2. Drag from `Sales[OrderDate]` to `Dates[Date]`

## 📊 **PART 4: CREATING YOUR FIRST MEASURE**

### **Step 1: Create Total Sales Measure**
1. Go to **Report view** (first icon)
2. In the **Fields pane** (right side), find **Sales** table
3. **Right-click** on Sales table
4. Select **New measure**

![New Measure Location](https://learn.microsoft.com/en-us/power-bi/transform-model/media/desktop-measures/measures-new-measure.png)

### **Step 2: Enter the Formula**
1. Formula bar appears at top
2. Type exactly:
```dax
Total Sales = SUM(Sales[SalesAmount])
```
3. Press **Enter** or click ✓

### **Step 3: Test Your Measure**
1. From **Visualizations pane** (right side), click **Card visual** icon
2. Drag your new **Total Sales** measure into the card
3. You should see a number appear!

![Card Visual](https://learn.microsoft.com/en-us/power-bi/create-reports/media/power-bi-visualization-card/pbi-create-card-small.png)

## 🔧 **PART 5: DAX USE-CASES PRACTICE**

### **Case 1: Calculated Column**
1. In **Data view** (second icon)
2. Select **Sales** table
3. Click **Table tools** → **New Column**
4. In formula bar, type:
```dax
Profit Margin % = 
DIVIDE(
    Sales[Profit],
    Sales[SalesAmount],
    0
) * 100
```
5. Press **Enter**

### **Case 2: Create More Measures**
Follow same steps as "Total Sales" to create:
```dax
Total Profit = SUM(Sales[Profit])
Total Cost = SUM(Sales[Cost])
Avg Order Value = AVERAGE(Sales[SalesAmount])
Number of Orders = COUNTROWS(Sales)
```

### **Case 3: RLS (Row-Level Security)**
1. Click **Modeling tab** in ribbon
2. Click **Manage Roles**

![Manage Roles Button](https://docs.microsoft.com/en-us/power-bi/admin/media/service-admin-rls/rls-modeling-tab.png)

3. Click **Create**
4. Type role name: `South Region Only`
5. For Locations table, click **DAX filter** box
6. Type:
```dax
[Region] = "South"
```
7. Click **Save**

**Test RLS:**
1. Click **Modeling tab** → **View as Roles**
2. Check your new role
3. Click **OK**
4. Your data now filters to South region only!

## 🎯 **PART 7: MASTERING CALCULATE**

### **Step 1: Create Basic CALCULATE Measure**
1. New measure in Sales:
```dax
Sales in Lagos = 
CALCULATE(
    [Total Sales],
    Locations[City] = "Lagos"
)
```

### **Step 2: Test with Visual**
1. From Visualizations, click **Table visual** icon
2. Drag `Locations[City]` to Rows
3. Drag both `[Total Sales]` and `[Sales in Lagos]` to Values
4. Observe: Lagos shows same in both, other cities show different!

### **Step 3: Using ALL Function**
```dax
% of Total = 
DIVIDE(
    [Total Sales],
    CALCULATE(
        [Total Sales],
        ALL(Sales)
    )
)
```

### **Step 4: Visual Demonstration**
1. Create a **Slicer**:
   - Click Slicer icon in Visualizations
   - Drag `Locations[Region]` to slicer
2. Create a **Card** with `[% of Total]`
3. Select different regions - watch percentage change!

---

## 📝 **PART 8: USING VARIABLES**

### **Step 1: Create Variable Measure**
```dax
Profit Margin = 
VAR TotalSales = SUM(Sales[SalesAmount])
VAR TotalProfit = SUM(Sales[Profit])
RETURN
    DIVIDE(TotalProfit, TotalSales, 0)
```

### **Step 2: Complex Example**
```dax
Customer Category = 
VAR CustSales = SUM(Sales[SalesAmount])
VAR AvgSales = 
    CALCULATE(
        AVERAGEX(VALUES(Customers[CustomerID]), [Total Sales]),
        ALL(Customers)
    )
RETURN
    SWITCH(
        TRUE(),
        CustSales > AvgSales * 1.5, "High Value",
        CustSales > AvgSales, "Medium Value",
        "Low Value"
    )
```

## 🔍 **PART 10: TROUBLESHOOTING COMMON ISSUES**

### **Issue 1: "Cannot find table/column"**
**Solution:**
1. Check spelling exactly
2. Column names are case-sensitive
3. Use auto-complete: Type `S` then press Tab

### **Issue 2: "The syntax for '=' is incorrect"**
**Solution:**
```dax
-- WRONG:
Sales Lagos = CALCULATE([Total Sales], Locations[City] = "Lagos")

-- RIGHT:
Sales Lagos = 
CALCULATE(
    [Total Sales],
    Locations[City] = "Lagos"
)
```

### **Issue 3: "A circular dependency was detected"**
**Solution:**
1. Check for self-references
2. Avoid calculated columns based on measures that reference same table
3. Simplify your DAX

### **Issue 4: Visual Not Updating**
**Solution:**
1. Click **Refresh** button in Home tab
2. Check filters in Filters pane
3. Verify relationships in Model view
